```python
import json
import requests

url = "https://ects-tracker.onrender.com/ects-stand"
with open("../beispiel_anfrage.json", encoding="utf-8") as f:
    anfrage = json.load(f)

resp = requests.post(url, json=anfrage, timeout=120)
data = resp.json()
print(f"Total: {data['total_ects']} ECTS, Rest bis Zulassung: {data['rest_bis_zulassung']}")
for g in data["gruppen"]:
    print(f"{g['gruppe']}: {g['ist']}/{g['minimum']}, Restbedarf {g['restbedarf']}")
```

```python
# Erweiterung 1.2: Konfliktwarnung
with open("../beispiel_konflikt.json", encoding="utf-8") as f:
    anfrage = json.load(f)

resp = requests.post(url, json=anfrage, timeout=120)
for k in resp.json()["konflikte"]:
    print(f"{k['tag']} {k['von']}-{k['bis']}: {k['modul_a']} / {k['modul_b']}")
```
